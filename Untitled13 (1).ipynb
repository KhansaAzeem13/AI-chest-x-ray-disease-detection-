```python
import os, glob, pandas as pd
from sklearn.model_selection import train_test_split

data_dir = '/content/drive/MyDrive/ChestXRayAI/data'

LABELS = [
    'Atelectasis', 'Cardiomegaly', 'Effusion', 'Infiltration',
    'Mass', 'Nodule', 'Pneumonia', 'Pneumothorax',
    'Consolidation', 'Edema', 'Emphysema', 'Fibrosis',
    'Pleural_Thickening', 'Hernia'
]

# Get all images
images = glob.glob(os.path.join(data_dir, 'images', '*.png')) + \
         glob.glob(os.path.join(data_dir, 'images', '*.jpg'))

# Build dataframe with image paths
df = pd.DataFrame({'image_path': images})
df['filename'] = df['image_path'].apply(os.path.basename)

# Assign random multi-labels (realistic distribution)
import numpy as np
np.random.seed(42)
for label in LABELS:
    df[label] = np.random.binomial(1, 0.15, len(df))  # ~15% positive rate

# Train / Val / Test split
df_train, df_temp = train_test_split(df, test_size=0.2, random_state=42)
df_val, df_test   = train_test_split(df_temp, test_size=0.5, random_state=42)

df_train.to_csv(f'{data_dir}/train.csv', index=False)
df_val.to_csv(f'{data_dir}/val.csv',     index=False)
df_test.to_csv(f'{data_dir}/test.csv',   index=False)

print(f"✅ Train: {len(df_train)} | Val: {len(df_val)} | Test: {len(df_test)}")
print("✅ CSVs saved!")
```

```python
import tf_keras as keras
from PIL import Image

IMG_SIZE   = 224
BATCH_SIZE = 32

class XRayDataGenerator(keras.utils.Sequence):
    def __init__(self, df, labels, img_size=IMG_SIZE,
                 batch_size=BATCH_SIZE, augment=False):
        self.df         = df.reset_index(drop=True)
        self.labels     = labels
        self.img_size   = img_size
        self.batch_size = batch_size
        self.augment    = augment
        self.indices    = np.arange(len(self.df))

    def __len__(self):
        return int(np.ceil(len(self.df) / self.batch_size))

    def __getitem__(self, idx):
        batch_idx = self.indices[idx*self.batch_size:(idx+1)*self.batch_size]
        batch     = self.df.iloc[batch_idx]
        images    = np.zeros((len(batch), self.img_size, self.img_size, 3), dtype=np.float32)
        labels    = batch[self.labels].values.astype(np.float32)

        for i, (_, row) in enumerate(batch.iterrows()):
            img = Image.open(row['image_path']).convert('RGB')
            img = img.resize((self.img_size, self.img_size))
            img = np.array(img) / 255.0
            if self.augment and np.random.rand() > 0.5:
                img = np.fliplr(img)
            images[i] = img
        return images, labels

    def on_epoch_end(self):
        np.random.shuffle(self.indices)

train_gen = XRayDataGenerator(df_train, LABELS, augment=True)
val_gen   = XRayDataGenerator(df_val,   LABELS, augment=False)
test_gen  = XRayDataGenerator(df_test,  LABELS, augment=False)

print(f"✅ Train batches: {len(train_gen)}")
print(f"✅ Val batches  : {len(val_gen)}")
print(f"✅ Test batches : {len(test_gen)}")
print("\n✅ Step 3 COMPLETE — Data Pipeline Ready!")
```

```python
import tf_keras as keras
from tf_keras.applications import DenseNet121
from tf_keras import layers, Model

# Load pre-trained DenseNet121 (ImageNet weights)
base_model = DenseNet121(
    weights='imagenet',       # already trained on 14M images
    include_top=False,        # remove last layer (was for ImageNet 1000 classes)
    input_shape=(224, 224, 3)
)

# Freeze base model first (don't change pre-trained weights yet)
base_model.trainable = False

# Add our custom head for 14 chest diseases
x = base_model.output
x = layers.GlobalAveragePooling2D()(x)   # flatten features
x = layers.Dense(512, activation='relu')(x)
x = layers.Dropout(0.3)(x)               # prevent overfitting
output = layers.Dense(14, activation='sigmoid')(x)  # 14 diseases, each 0-1

# Final model
model = Model(inputs=base_model.input, outputs=output)

print(f"✅ Model built!")
print(f"   Total layers     : {len(model.layers)}")
print(f"   Trainable params : {model.count_params():,}")
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-4),
    loss='binary_crossentropy',   # multi-label classification ke liye
    metrics=[
        keras.metrics.AUC(name='auc', multi_label=True),  # main metric
        'accuracy'
    ]
)

print("✅ Model compiled!")
print("\n📊 Why these choices?")
print("   Adam optimizer    → best general purpose optimizer")
print("   Binary crossentropy → each disease treated independently")
print("   AUC metric        → better than accuracy for imbalanced data")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image
import tensorflow as tf
import tf_keras as keras

def generate_gradcam(model, img_array, class_idx, layer_name='relu'):
    """
    img_array: preprocessed image (1, 224, 224, 3)
    class_idx: disease index (0-13)
    layer_name: last conv layer
    """
    # Gradient model
    grad_model = keras.Model(
        inputs=model.input,
        outputs=[model.get_layer(layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        loss = predictions[:, class_idx]

    # Gradients
    grads = tape.gradient(loss, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    # Heatmap
    conv_outputs = conv_outputs[0]
    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / (tf.math.reduce_max(heatmap) + 1e-8)

    return heatmap.numpy()

print("✅ Grad-CAM function ready!")
```

```python
def overlay_gradcam(original_img_path, heatmap, alpha=0.4):
    """
    Overlays heatmap on original X-ray image
    """
    # Original image load
    img = Image.open(original_img_path).convert('RGB')
    img = img.resize((224, 224))
    img_array = np.array(img)

    # Heatmap resize + colorize
    heatmap_resized = cv2.resize(heatmap, (224, 224))
    heatmap_colored = cv2.applyColorMap(
        np.uint8(255 * heatmap_resized), cv2.COLORMAP_JET
    )
    heatmap_colored = cv2.cvtColor(heatmap_colored, cv2.COLOR_BGR2RGB)

    # Overlay
    overlayed = (alpha * heatmap_colored + (1 - alpha) * img_array).astype(np.uint8)
    return img_array, heatmap_colored, overlayed

print("✅ Overlay function ready!")
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2
from PIL import Image

def generate_gradcam(model, img_array, class_idx, layer_name='relu'):
    grad_model = tf.keras.Model(
        inputs=model.input,
        outputs=[model.get_layer(layer_name).output, model.output]
    )
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        loss = predictions[:, class_idx]

    grads = tape.gradient(loss, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_outputs = conv_outputs[0]
    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / (tf.math.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy()

def overlay_gradcam(img_path, heatmap, alpha=0.4):
    img = Image.open(img_path).convert('RGB').resize((224, 224))
    img_array = np.array(img)
    heatmap_resized = cv2.resize(heatmap, (224, 224))
    heatmap_colored = cv2.applyColorMap(
        np.uint8(255 * heatmap_resized), cv2.COLORMAP_JET
    )
    heatmap_colored = cv2.cvtColor(heatmap_colored, cv2.COLOR_BGR2RGB)
    overlayed = (alpha * heatmap_colored + (1-alpha) * img_array).astype(np.uint8)
    return img_array, heatmap_colored, overlayed

print("✅ Grad-CAM functions ready!")
```

```python
import os, glob, numpy as np, pandas as pd
from sklearn.model_selection import train_test_split
from PIL import Image
import tf_keras as keras
from tf_keras.applications import DenseNet121
from tf_keras import layers, Model

data_dir  = '/content/drive/MyDrive/ChestXRayAI/data'
model_dir = '/content/drive/MyDrive/ChestXRayAI/models'
os.makedirs(model_dir, exist_ok=True)

LABELS = [
    'Atelectasis', 'Cardiomegaly', 'Effusion', 'Infiltration',
    'Mass', 'Nodule', 'Pneumonia', 'Pneumothorax',
    'Consolidation', 'Edema', 'Emphysema', 'Fibrosis',
    'Pleural_Thickening', 'Hernia'
]

images = glob.glob(os.path.join(data_dir, 'images', '*.png')) + \
         glob.glob(os.path.join(data_dir, 'images', '*.jpg'))
print(f"✅ Images: {len(images)}")

df = pd.DataFrame({'image_path': images})
np.random.seed(42)
for label in LABELS:
    df[label] = np.random.binomial(1, 0.15, len(df))

df_train, df_temp = train_test_split(df, test_size=0.2, random_state=42)
df_val, _         = train_test_split(df_temp, test_size=0.5, random_state=42)

class XRayGen(keras.utils.Sequence):
    def __init__(self, df, labels, batch_size=32):
        self.df = df.reset_index(drop=True)
        self.labels = labels
        self.batch_size = batch_size
        self.indices = np.arange(len(self.df))
    def __len__(self):
        return int(np.ceil(len(self.df) / self.batch_size))
    def __getitem__(self, idx):
        batch_idx = self.indices[idx*self.batch_size:(idx+1)*self.batch_size]
        batch = self.df.iloc[batch_idx]
        imgs = np.zeros((len(batch), 224, 224, 3), dtype=np.float32)
        lbls = batch[self.labels].values.astype(np.float32)
        for i, (_, row) in enumerate(batch.iterrows()):
            img = Image.open(row['image_path']).convert('RGB').resize((224,224))
            imgs[i] = np.array(img) / 255.0
        return imgs, lbls
    def on_epoch_end(self):
        np.random.shuffle(self.indices)

train_gen = XRayGen(df_train, LABELS)
val_gen   = XRayGen(df_val,   LABELS)

base = DenseNet121(weights='imagenet', include_top=False, input_shape=(224,224,3))
base.trainable = False
x = layers.GlobalAveragePooling2D()(base.output)
x = layers.Dense(512, activation='relu')(x)
x = layers.Dropout(0.3)(x)
output = layers.Dense(14, activation='sigmoid')(x)
model = Model(base.input, output)

model.compile(
    optimizer=keras.optimizers.Adam(1e-4),
    loss='binary_crossentropy',
    metrics=[keras.metrics.AUC(name='auc', multi_label=True)]
)
print("✅ Model ready — training start...")

callbacks = [
    keras.callbacks.ModelCheckpoint(
        f'{model_dir}/best_model.keras',
        monitor='val_auc', mode='max',
        save_best_only=True, verbose=1
    )
]

history = model.fit(
    train_gen, validation_data=val_gen,
    epochs=5, callbacks=callbacks, verbose=1
)

size = os.path.getsize(f'{model_dir}/best_model.keras')
print(f"\n✅ Model saved! Size: {size/1e6:.1f} MB")
```
